Using Linear Regression

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
import tensorflow as tf 
```

```python
x_train = np.array([[1.0],[2.0]], dtype=np.float32)
y_train = np.array([[300.0],[500.0]],dtype=np.float32) 
```

```python
linear_layer = tf.keras.layers.Dense(units=1, activation='linear')
```

```python
#Input to the layer must be 2-D, so we'll reshape it
a1 = linear_layer(x_train[0].reshape(1,1))
```

```python
print(a1)
```

```python
#Wieghts are randomly initialized to small numbers and the bias default to beign initialized to zero
w, b = linear_layer.get_weights()
print("w:",w," b:",b)
```

```python
#The weights are initialized to random values so let's set them to some known values
set_w = np.array([[200]])
set_b = np.array([100])

#Set weights takes a list of numpy arrays
linear_layer.set_weights([set_w, set_b])
print(linear_layer.get_weights())
```

```python
#Let's compare euqation (1) to the layer output
a1 = linear_layer(x_train[0].reshape(1,1))
print(a1)
alin = np.dot(set_w, x_train[0].reshape(1,1)) + set_b
print(alin)
```

```python
#Use our linear layer to make predictions on our training data
prediction_tf = linear_layer(x_train)
prediction_np = np.dot(x_train, set_w) + set_b
```

```python
print(prediction_tf)
print(prediction_np)
```

Using Logistic Regression

```python
#data set
x_train = np.array([0.,1,2,3,4],dtype=np.float32).reshape(-1,1) #2-D Matrix
y_train = np.array([0,0,0,1,1,1], dtype=np.float32).reshape(-1,1) # 2-D matrix 
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.losses import MeanSquaredError, BinaryCrossentropy
from tensorflow.keras.activations import sigmoid
```

```python
#logistic neuron
model = Sequential(
    [
        tf.keras.layers.Dense(1, input_dim=1, activation='sigmoid', name='L1')
    ]
)
```

```python
model.summary()
```

```python
#Initializing random weights to L1
logistic_layer = model.get_layer('L1')
w,b = logistic_layer.get_weights()
print(w,b)
print(w.shape, b.shape)
```

```python
#Setting some know weights and bias
set_w = np.array([[2]])
set_b = np.array([-4.5])

logistic_layer.set_weights([set_w, set_b])
print(logistic_layer.get_weights())
```

```python
#Comparing equation 2 to the layer output
a1 = model.predict(x_train[0].reshape(1,1))
print(a1)
alog = sigmoid(np.dot(set_w, x_train[0].reshape(1,1)) + set_b)
print(alog)
```
